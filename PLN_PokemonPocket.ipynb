```
from tcgdexsdk import TCGdex, Language, Query
from langchain_core.tools import Tool

tcgdex = TCGdex(Language.EN)

# Re-initialize listCards to ensure it's empty before re-populating
listCards = []

# imagemCards armazena as imagens das cartas
imagemCards = []

# Re-populating listCards correctly, ensuring only Card objects are added
# (Corrected cell 1hAa6ugzKg3M)

#Pegando as cartas do primeiro pacote Genetic Apex
for i in range(1,227):
  card_id = "A1-" + str(i).zfill(3)
  card = await tcgdex.card.get(card_id)
  if (card.description != None):
    listCards.append(card)
    imagemCards.append(card.get_image_url("low", "jpg"))

#Pegando as cartas com descrição do segundo pacote Mythical Island (FIXED: append card, not card.description)
for i in range(1,69):
  card_id = "A1a-" + str(i).zfill(3)
  card = await tcgdex.card.get(card_id)
  if (card.description != None):
    listCards.append(card)
    imagemCards.append(card.get_image_url("low", "jpg"))

#Pegando as cartas com descrição do terceiro pacote Space-Time Smackdown
for i in range(1,156):
  card_id = "A2-" + str(i).zfill(3)
  card = await tcgdex.card.get(card_id)
  if (card.description != None):
    listCards.append(card)
    imagemCards.append(card.get_image_url("low", "jpg"))

#Pegando as cartas com descrição do quarto pacote Triumphant Light
for i in range(1,76):
  card_id = "A2a-" + str(i).zfill(3)
  card = await tcgdex.card.get(card_id)
  if (card.description != None):
    listCards.append(card)
    imagemCards.append(card.get_image_url("low", "jpg"))

#Pegando as cartas com descrição do quinto pacote Shinning Revelry
for i in range(1,73):
  card_id = "A2b-" + str(i).zfill(3)
  card = await tcgdex.card.get(card_id)
  if (card.description != None):
    listCards.append(card)
    imagemCards.append(card.get_image_url("low", "jpg"))

#Pegando as cartas com descrição do sexto pacote Celestial Guardians
for i in range(1,156):
  card_id = "A3-" + str(i).zfill(3)
  card = await tcgdex.card.get(card_id)
  if (card.description != None):
    listCards.append(card)
    imagemCards.append(card.get_image_url("low", "jpg"))

#Pegando as cartas com descrição do setimo pacote Extradimensional Crisis
for i in range(1,70):
  card_id = "A3a-" + str(i).zfill(3)
  card = await tcgdex.card.get(card_id)
  if (card.description != None):
    listCards.append(card)
    imagemCards.append(card.get_image_url("low", "jpg"))

#Pegando as cartas com descrição do oitavo pacote Eevee Grove
for i in range(1,70):
  card_id = "A3b-" + str(i).zfill(3)
  card = await tcgdex.card.get(card_id)
  if (card.description != None):
    listCards.append(card)
    imagemCards.append(card.get_image_url("low", "jpg"))

#Pegando as cartas com descrição do nono pacote Wisdom of Sea and Sky
for i in range(1,162):
  card_id = "A4-" + str(i).zfill(3)
  card = await tcgdex.card.get(card_id)
  if (card.description != None):
    listCards.append(card)
    imagemCards.append(card.get_image_url("low", "jpg"))

#Pegando as cartas com descrição do decimo pacote Secluded Springs
for i in range(1,72):
  card_id = "A4a-" + str(i).zfill(3)
  card = await tcgdex.card.get(card_id)
  if (card.description != None):
    listCards.append(card)
    imagemCards.append(card.get_image_url("low", "jpg"))

print(f"Total de cartas com descrição na listCards: {len(listCards)}")

# Re-define the formatar_card_para_texto function to ensure it's in scope
def formatar_card_para_texto(card_objeto):
    """Extrai informações relevantes do objeto Card e formata como string."""
    # Ensure card_objeto is a Card object, not a string (if that was an issue from earlier errors)
    if not hasattr(card_objeto, 'name'):
        # If for some reason a non-Card object made it here, handle it gracefully
        return f"--- INFORMAÇÕES INVÁLIDAS --- Objeto não é um Card: {card_objeto}"

    info_card = f"""
--- INFORMAÇÕES DO POKÉMON ---
Nome: {card_objeto.name}
ID: {card_objeto.id}
Categoria: {card_objeto.category}
HP: {card_objeto.hp}
Tipos: {', '.join(card_objeto.types)}
Estágio: {card_objeto.stage}
Descrição: {card_objeto.description}

Ataques:
"""

    if card_objeto.attacks:
        for ataque in card_objeto.attacks:
            cost_str = ', '.join(ataque.cost) if isinstance(ataque.cost, list) else 'N/A'
            info_card += f"- {ataque.name}: Dano: {ataque.damage if ataque.damage else 'N/A'}, Custo: {cost_str}\n"
    else:
        info_card += "- Nenhum ataque listado.\n"

    if card_objeto.weaknesses and len(card_objeto.weaknesses) > 0:
        info_card += f"\nFraquezas: {card_objeto.weaknesses[0].type} ({card_objeto.weaknesses[0].value})"
    else:
        info_card += "Nenhuma fraqueza listada."

    return info_card


print("TCGdex initialized, listCards re-populated successfully.")
```

## Criação da ferramenta `search_pokemon_by_name` para Sumarização

Modificar a função existente `search_pokemon_by_name` (ou criar uma nova ferramenta) para que ela utilize a `formatar_card_para_texto` e a `listCards` para buscar informações detalhadas de um Pokémon da API do TCGdex. A saída desta ferramenta deve ser uma string formatada e rica em detalhes, ideal para o LLM realizar a sumarização.


```
from langchain_core.tools import Tool

def _get_pokemon_details(pokemon_name: str) -> str:
    """Busca informações detalhadas de um Pokémon pelo nome na listCards.
    Retorna uma string formatada com os detalhes do Pokémon ou uma mensagem de erro.
    """
    for card in listCards:
        if hasattr(card, 'name') and card.name.lower() == pokemon_name.lower():
            return formatar_card_para_texto(card)
    return f"Pokémon '{pokemon_name}' não encontrado na lista."

# Criar a ferramenta LangChain
search_pokemon_by_name = Tool(
    name="search_pokemon_by_name",
    func=_get_pokemon_details,
    description="""Use esta ferramenta para buscar informações detalhadas de um Pokémon pelo nome.
    A entrada deve ser o nome exato do Pokémon (string). Esta ferramenta é ideal para coletar dados
    para sumarização e análise de características de combate, pontos fortes e fracos.
    """
)

print("Tool 'search_pokemon_by_name' created successfully.")
```

## Criação da ferramenta `analyze_pokemon_description_sentiment` para fazer classificação de texto

```
from langchain_core.tools import Tool

def _analyze_sentiment(text: str) -> str:
    """Analisa o sentimento de um texto (descrição de Pokémon) usando o LLM."""
    # Aqui, a ferramenta usa o próprio LLM para realizar a PLN de Sentimento
    sentiment_prompt = f"""
    Analise o seguinte texto e classifique o sentimento geral da descrição do Pokémon como:
    'Agressivo', 'Defensivo', 'Misterioso', 'Pacífico', 'Brincalhão', 'Neutro' ou 'Outro'.
    Explique brevemente sua classificação.

    Texto: {text}
    Sentimento:
    """
    # Assumindo que 'llm' está disponível globalmente ou passado para a função
    response = llm.invoke(sentiment_prompt)
    return response.content

analyze_pokemon_description_sentiment = Tool(
    name="analyze_pokemon_description_sentiment",
    func=_analyze_sentiment,
    description="""Use esta ferramenta para analisar o sentimento (positivo, negativo, neutro, agressivo, pacífico, brincalhão, etc.) de um texto.
    A entrada deve ser a descrição do Pokémon (string). Esta ferramenta é útil para entender a natureza ou disposição do Pokémon.
    """
)
```

## Configurar o Agente LangChain para Sumarização

Desenvolver o agente LangChain, utilizando `ChatGoogleGenerativeAI` com o modelo 'gemini-2.5-flash' e a ferramenta `search_pokemon_by_name`, com um prompt que instrui o agente a gerar um resumo conciso sobre suas características de combate, pontos fortes e fracos de um Pokémon.


```
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import AgentExecutor, create_tool_calling_agent
from langchain_core.messages import HumanMessage, BaseMessage
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

# 2. Inicialize o modelo de linguagem
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")

# 3. Defina as ferramentas que o agente utilizará (search_pokemon_by_name já está definida)
tools = [search_pokemon_by_name, analyze_pokemon_description_sentiment]

# 4. Crie um prompt para o agente
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "Você é um pesquisador de Pokémon. Você tem duas tarefas: uma delas é analisar as informações fornecidas sobre um Pokémon e gerar um resumo conciso de suas características de combate, pontos fortes e fracos. Para isso, use a ferramenta 'search_pokemon_by_name' para obter os detalhes do Pokémon. O resumo deve ser objetivo e focado nas capacidades de combate. A outra tarefa é analisar a descrição das cartas e classificar o sentimento que o pokémon transmite. Para esta tarefa, você também deve primeiro usar 'search_pokemon_by_name' para obter a descrição do Pokémon antes de usar a ferramenta de análise de sentimento."),
        MessagesPlaceholder("chat_history", optional=True),
        ("human", "{input}"),
        MessagesPlaceholder("agent_scratchpad"),
    ]
)

# 5. Crie o agente
agent = create_tool_calling_agent(llm, tools, prompt)

# 6. Instancie o AgentExecutor
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True)

print("LangChain agent and executor re-configured successfully.")
```

```
import google.generativeai as genai
import requests
from PIL import Image
from io import BytesIO
from google.api_core import exceptions
from IPython.display import Markdown, display

# LangChain imports
from langchain_core.tools import StructuredTool
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import AgentExecutor, create_tool_calling_agent
from langchain_core.messages import HumanMessage, BaseMessage
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from pydantic import BaseModel, Field

# --- LLM Initialization (for agent orchestration) ---
# Ensure llm is defined. Using gemini-2.5-flash for agent orchestration.
try:
    llm
except NameError:
    llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")

# --- Multimodal Model Definition for the Tool ---
# User requested gemini-2.0-flash for image analysis due to quota
MULTIMODAL_MODEL_NAME = 'gemini-2.0-flash'

def _analyze_image_with_llm_tool(image_url: str, original_description: str) -> str:
    """Analisa uma imagem de Pokémon e compara com uma descrição original,
    gerando uma nota e uma nova descrição se a original for insuficiente.
    Retorna a resposta formatada do LLM."""
    try:
        # Load the image
        response_img = requests.get(image_url)
        img = Image.open(BytesIO(response_img.content))

        # Initialize the multimodal model within the tool for robustness
        multimodal_model = genai.GenerativeModel(MULTIMODAL_MODEL_NAME)

        prompt_for_image = f"""
Por favor, analise a imagem do card Pokémon fornecida e compare com a descrição fornecida e de uma nota de 0 a 10 para o quanto se encaixa com a imagem, se a nota for inferior a 6 faça uma nova descrição. A resposta deve seguir o formato abaixo:

Descrição original: {original_description} \n
Nota da descrição: [Nota da descrição enviada] \n

Descrição nova: [Nova descrição se a nota do original for menor que 6] \n
Nota da nova descrição: [Nota da nova descrição se a nota do original for menor que 6]
"""
        llm_response = multimodal_model.generate_content([prompt_for_image, img])
        # Removed the print statement that caused redundancy
        return llm_response.text
    except exceptions.TooManyRequests as e:
        return f"Erro: Quota da API excedida na ferramenta de análise de imagem. Por favor, aguarde e tente novamente em alguns segundos. Detalhes: {e}"
    except Exception as e:
        return f"Ocorreu um erro inesperado na ferramenta de análise de imagem: {e}"

# --- Arguments Schema for the StructuredTool ---
class AnalyzeImageInput(BaseModel):
    image_url: str = Field(description="URL da imagem do card Pokémon a ser analisada.")
    original_description: str = Field(description="Descrição textual original do Pokémon para comparação.")

# --- Define the LangChain StructuredTool ---
analyze_image_and_description_tool = StructuredTool(
    name="analyze_image_and_description",
    func=_analyze_image_with_llm_tool,
    description="""
    Use esta ferramenta para analisar uma imagem de um card Pokémon e compará-la com uma descrição textual.
    Ela retorna uma avaliação da descrição original e, se necessário, uma nova descrição da imagem.
    """,
    args_schema=AnalyzeImageInput
)

# --- Agent for Image Analysis ---
image_analysis_tools = [analyze_image_and_description_tool]

image_agent_prompt = ChatPromptTemplate.from_messages(
    [
        # Modified system prompt to instruct direct output of tool result
        ("system", "Você é um assistente. Sua única tarefa é usar a ferramenta 'analyze_image_and_description' para comparar a imagem de um card Pokémon com uma descrição fornecida pelo usuário e, em seguida, apresentar *apenas* o resultado direto fornecido pela ferramenta."),
        MessagesPlaceholder("chat_history", optional=True),
        ("human", "{input}"),
        MessagesPlaceholder("agent_scratchpad"),
    ]
)

image_analysis_agent = create_tool_calling_agent(llm, image_analysis_tools, image_agent_prompt)
image_analysis_agent_executor = AgentExecutor(agent=image_analysis_agent, tools=image_analysis_tools, verbose=False) # verbose=False to keep output clean

# --- Test the LangChain Agent (Synchronous call to AgentExecutor.invoke) ---
ncard = 93 # As given by the user in the original cell for Zapdos
image_url_to_analyze = imagemCards[ncard]
description_to_analyze = listCards[ncard].description
pokemon_name = listCards[ncard].name

print(f"\n--- Invocando o Agente para análise de imagem e descrição do card {pokemon_name} (LangChain) ---")

# Construct the input for the agent
agent_input_text = f"Analise a imagem do card Pokémon em {image_url_to_analyze} e compare com a descrição: '{description_to_analyze}'. Forneça uma nota para a descrição original e, se a nota for inferior a 6, gere uma nova descrição conforme o formato definido na ferramenta."

try:
    # AgentExecutor.invoke can be called synchronously; it handles internal async calls.
    response_agent = image_analysis_agent_executor.invoke({"input": agent_input_text})
    print("--- RESPOSTA DO AGENTE (Análise de Imagem e Descrição) ---")
    display(Markdown(response_agent["output"])) # Display as Markdown
except Exception as e:
    print(f"Ocorreu um erro ao invocar o agente de análise de imagem: {e}")
```
